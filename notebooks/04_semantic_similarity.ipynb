# 04 · Semantic similarity by domain
SBERT and rescaled BERTScore at T3, per domain and paraphraser.

```python
import os, sys
ROOT = os.path.abspath("..")          # notebooks/ -> repository root
sys.path.insert(0, ROOT)
os.chdir(ROOT)
import pandas as pd
pd.set_option("display.max_colwidth", 120)
```

```python
dom = pd.read_csv("experiments/baseline_results/domain_comparison.csv")
t3 = dom[(dom.generation == "T3") & (dom.paraphraser != "dipper(high)")]
t3.pivot(index="domain", columns="paraphraser", values="sbert_cosine").round(3)
```

```python
t3.pivot(index="domain", columns="paraphraser", values="bertscore_rescaled").round(3)
```
